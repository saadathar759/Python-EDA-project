```python
import pandas as pd

# 1. Load the dataset from the URL into a DataFrame named 'data'
url = "https://docs.google.com/spreadsheets/d/1zLGcNoattvIPSjkHYF9lzn6R_kGxGWqC/export?format=csv"
data = pd.read_csv(url)
```

```python
df = pd.DataFrame(data)
print(df)
```

```python
rename_columns = {
    'InvoiceNo': 'transaction_id',
    'Description': 'item_name',
    'Quantity': 'order_quantity',
    'UnitPrice': 'item_price',
    'InvoiceDate': 'transaction_date'
}

df.rename(columns=rename_columns, inplace=True)

print(df.columns)
```
